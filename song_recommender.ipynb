```python
import pandas as pd
import numpy as np
```

```python
songs=pd.read_csv("spotify_songs_dataset.csv")
songs.head()
```

```python
songs['composer'].nunique()
```

```python
songs.isnull().mean()
```

```python
songs=songs[['song_id','song_title','artist','album','genre','explicit_content']]
```

```python
songs.head()
```

```python
songs['artist']=songs['artist'].str.replace(" ","")
songs.head()
```

```python
songs['album']=songs['album'].str.replace(".","")
songs['song_title']=songs['song_title'].str.replace(".","")
```

```python
songs['tags']=songs[['artist','album','genre','explicit_content']].agg(" ".join, axis=1)
songs.head(50000)
```

```python
songs['tags']=songs['tags'].apply(lambda x:x.lower())
```

```python
songs.head()
```

```python
songs_final=songs[['song_id','song_title','tags','artist']]
songs_final.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=5000,stop_words='english')
vectors=cv.fit_transform(songs_final['tags']).toarray()
vectors
```

```python
cv.get_feature_names_out()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity=cosine_similarity(vectors)
similarity
```

```python
similarity[0]
```

```python
def song(name):
    if len(songs_final[songs_final['song_title']==name])==1:
        song_index=songs_final[songs_final['song_title']==name].index[0]
        songs_list=sorted(list(enumerate(similarity[song_index])),key=lambda x:x[1],reverse=True)[1:11]
        for x in songs_list:
            print(songs_final.iloc[x[0]].song_title)
    elif len(songs_final[songs_final['song_title']==name])>1:
        artist=input("Enter the name of the artist")
        song_index=songs_final[(songs_final['song_title']==name) & (songs_final['artist']==artist)].index[0]
        songs_list=sorted(list(enumerate(similarity[song_index])),key=lambda x:x[1],reverse=True)[1:11]
        for x in songs_list:
            print(songs_final.iloc[x[0]].song_title)
    else:
        print("Invalid Song")
```
